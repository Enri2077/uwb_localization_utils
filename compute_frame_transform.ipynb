```python
import yaml
import math
import os

with open(os.path.expanduser("~/tmp/anchors.yaml")) as f:
    a = yaml.safe_load(f)

a1_x_map = a['A1']['map']['x']
a1_y_map = a['A1']['map']['y']

a2_x_map = a['A2']['map']['x']
a2_y_map = a['A2']['map']['y']

a3_x_map = a['A3']['map']['x']
a3_y_map = a['A3']['map']['y']

a4_x_map = a['A4']['map']['x']
a4_y_map = a['A4']['map']['y']

print("\n  Map Frame:")
print(f"a1_x_map = {a1_x_map: >6.2f}    a1_y_map = {a1_y_map: >6.2f}")
print(f"a2_x_map = {a2_x_map: >6.2f}    a2_y_map = {a2_y_map: >6.2f}")
print(f"a3_x_map = {a3_x_map: >6.2f}    a3_y_map = {a3_y_map: >6.2f}")
print(f"a4_x_map = {a4_x_map: >6.2f}    a4_y_map = {a4_y_map: >6.2f}")

a1_x_uwb = a['A1']['uwb_frame']['x']
a1_y_uwb = a['A1']['uwb_frame']['y']

a2_x_uwb = a['A2']['uwb_frame']['x']
a2_y_uwb = a['A2']['uwb_frame']['y']

a3_x_uwb = a['A3']['uwb_frame']['x']
a3_y_uwb = a['A3']['uwb_frame']['y']

a4_x_uwb = a['A4']['uwb_frame']['x']
a4_y_uwb = a['A4']['uwb_frame']['y']

print("\n  UWB Frame:")
print(f"a1_x_uwb = {a1_x_uwb: >6.2f}    a1_y_uwb = {a1_y_uwb: >6.2f}")
print(f"a2_x_uwb = {a2_x_uwb: >6.2f}    a2_y_uwb = {a2_y_uwb: >6.2f}")
print(f"a3_x_uwb = {a3_x_uwb: >6.2f}    a3_y_uwb = {a3_y_uwb: >6.2f}")
print(f"a4_x_uwb = {a4_x_uwb: >6.2f}    a4_y_uwb = {a4_y_uwb: >6.2f}")

assert(a1_x_uwb == 0.0)
assert(a1_y_uwb == 0.0)
assert(a2_y_uwb == 0.0)

yaw = math.atan2(a2_y_map - a1_y_map, a2_x_map - a1_x_map)
x = a1_x_map
y = a1_y_map

print("\n  Transform map -> uwb_frame:")
print(f"x={x:.3f}  y={y:.3f} yaw={yaw:.5f}")
```
